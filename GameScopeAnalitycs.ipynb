```python

import requests
import pandas as pd
import time

BASE_URL = "https://steamspy.com/api.php"

# Classements à récupérer
classements = [
    "top100in2weeks",
    "top100forever",
    "top100owned"
]

catalogues = []

for classement in classements:
    print(f"Récupération : {classement}")

    try:
        response = requests.get(
            BASE_URL,
            params={"request": classement},
            timeout=30
        )
        response.raise_for_status()

        donnees = response.json()
        catalogue = pd.DataFrame.from_dict(
            donnees, orient="index"
        )

        if not catalogue.empty:
            catalogue["appid"] = pd.to_numeric(
                catalogue["appid"], errors="coerce"
            )
            catalogue = catalogue.dropna(subset=["appid"])
            catalogue["appid"] = catalogue["appid"].astype(int)
            catalogue["source_classement"] = classement
            catalogues.append(catalogue)

            print(f"  {len(catalogue)} jeux récupérés")
        else:
            print("  Aucun jeu retourné")

    except (requests.RequestException, ValueError) as erreur:
        print(f"  Erreur : {erreur}")

    # Pause pour éviter d'enchaîner les requêtes
    time.sleep(65)

if catalogues:
    jeux = pd.concat(catalogues, ignore_index=True)

    # Fusion des doublons en conservant une ligne par jeu
    jeux = jeux.drop_duplicates(subset="appid", keep="first")

    print("\nNombre de jeux uniques :", len(jeux))
    print("\nColonnes disponibles :")
    print(jeux.columns.tolist())

    display(jeux.head(10))
else:
    print("Aucun catalogue récupéré.")
```

```python
print("Nombre de jeux :", len(jeux))
print("Nombre d'identifiants uniques :", jeux["appid"].nunique())

display(
    jeux[["appid", "name", "owners", "ccu"]]
    .head(20)
)
```

```python
print("Nombre de jeux uniques :", jeux["appid"].nunique())

print("\nRépartition des sources :")
print(jeux["source_classement"].value_counts())

print("\nExemples de jeux :")
display(jeux[["appid", "name", "owners", "ccu"]].head(10))
```

```python
params = {
    "request": "all",
    "page": 1
}

response = requests.get(
    "https://steamspy.com/api.php",
    params=params,
    timeout=30
)

print("Code HTTP :", response.status_code)

if response.status_code == 200:
    donnees_page2 = response.json()
    catalogue_page2 = pd.DataFrame.from_dict(
        donnees_page2, orient="index"
    )

    print("Nombre de jeux page 2 :", len(catalogue_page2))
    print(
        "Nombre d'identifiants déjà présents dans la page 1 :",
        catalogue_page2["appid"].isin(catalogue["appid"]).sum()
    )

    display(catalogue_page2.head())
else:
    print("Erreur :", response.text[:500])
```

```python

# Fusion des deux pages SteamSpy
catalogue_complet = pd.concat(
    [catalogue, catalogue_page2],
    ignore_index=True
)

# Suppression des éventuels doublons
catalogue_complet = catalogue_complet.drop_duplicates(
    subset="appid"
)

print("Nombre total de jeux :", len(catalogue_complet))
print(
    "Identifiants uniques :",
    catalogue_complet["appid"].nunique()
)

display(
    catalogue_complet[
        ["appid", "name", "owners", "positive", "negative", "ccu"]
    ].head(10)
)
```

```python
print("Valeurs manquantes :")
display(catalogue_complet.isna().sum())

print("\nTypes de données :")
catalogue_complet.info()

print("\nRépartition des propriétaires estimés :")
display(catalogue_complet["owners"].value_counts().head(10))
```

```python
import time

echantillon = catalogue_complet.head(10).copy()
details_jeux = []

for _, jeu in echantillon.iterrows():
    params = {
        "request": "appdetails",
        "appid": int(jeu["appid"])
    }

    try:
        response = requests.get(
            "https://steamspy.com/api.php",
            params=params,
            timeout=30
        )
        response.raise_for_status()
        details = response.json()

        details_jeux.append({
            "appid": int(jeu["appid"]),
            "genre": details.get("genre"),
            "tags": details.get("tags")
        })

        print("OK :", jeu["name"])

    except (requests.RequestException, ValueError) as erreur:
        print("Erreur pour", jeu["name"], ":", erreur)

    time.sleep(65)

df_genres_test = pd.DataFrame(details_jeux)

display(df_genres_test)
```

```python
print("Nombre de jeux enrichis :", len(df_genres_test))

print("\nGenres récupérés :")
display(
    df_genres_test[["appid", "genre"]]
)

print("\nGenres manquants :")
print(df_genres_test["genre"].isna().sum())

print("\nNombre de jeux avec des tags :")
print(df_genres_test["tags"].notna().sum())
```

```python

# Repérer les produits dont les tags suggèrent un DLC,
# un logiciel ou un outil plutôt qu'un jeu classique.

motifs_exclusion = [
    "DLC",
    "Software",
    "Utilities",
    "Video Production",
    "Audio Production",
    "Web Publishing",
    "Design & Illustration"
]

for motif in motifs_exclusion:
    masque = df["tags"].fillna("").str.contains(
        motif, case=False, regex=False
    )
    print(f"\n--- {motif} : {masque.sum()} résultats ---")
    display(df.loc[masque, ["appid", "name", "genre", "tags"]].head(10))
```

```python
# Repérer les produits clairement orientés logiciel ou outils
masque_logiciels = df["genre"].fillna("").str.contains(
    "Animation & Modeling|Audio Production|Video Production|"
    "Web Publishing|Utilities|Software Training",
    case=False,
    regex=True
)

candidats_a_verifier = df.loc[
    masque_logiciels,
    ["appid", "name", "genre", "tags"]
].copy()

display(candidats_a_verifier)
print("Produits à vérifier :", len(candidats_a_verifier))
```

```python

# Créer une copie de travail sans les produits identifiés
# comme logiciels ou outils.

df_nettoye = df.loc[~masque_logiciels].copy()

print("Catalogue initial :", len(df))
print("Catalogue après ce premier filtre :", len(df_nettoye))
print("Produits exclus :", len(df) - len(df_nettoye))

# Vérifier les genres encore manquants
print(
    "Genres manquants restants :",
    df_nettoye["genre"].isna().sum()
)

# Exporter le catalogue de travail
df_nettoye.to_csv(
    "gamescope_catalogue_jeux.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Fichier créé : gamescope_catalogue_jeux.csv")
```

```python
import pandas as pd

df = pd.read_csv("gamescope_catalogue_jeux.csv")

print("Nombre de lignes :", len(df))
print("Nombre de colonnes :", len(df.columns))

print("\nTypes des colonnes :")
print(df.dtypes)

print("\nValeurs manquantes :")
print(df.isna().sum())

print("\nDoublons d'identifiants :", df["appid"].duplicated().sum())
```

```python
# Valeurs manquantes par colonne
manquants = df.isna().sum()
print("Valeurs manquantes par colonne :")
print(manquants[manquants > 0].sort_values(ascending=False))

# Vérifier les valeurs de prix
print("\nExemples de prix :")
display(
    df[["name", "price", "initialprice", "discount"]].head(10)
)

# Statistiques des prix
print("\nStatistiques des prix :")
print(df[["price", "initialprice", "discount"]].describe())
```

```python
print("Exemples de propriétaires estimés :")
display(df["owners"].dropna().head(10))

print("\nValeurs de remise :")
print(sorted(df["discount"].dropna().unique()))

print("\nGenres les plus fréquents :")
print(df["genre"].value_counts().head(15))

print("\nNombre de jeux gratuits (prix nul) :")
print((df["price"] == 0).sum())
```

```python

# Créer des colonnes numériques utiles pour l'analyse

df["prix_euros"] = df["price"] / 100
df["prix_initial_euros"] = df["initialprice"] / 100

# Extraire les bornes de la fourchette d'owners
owners_bornes = (
    df["owners"]
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+)\s*\.\.\s*(\d+)")
)

df["owners_min"] = pd.to_numeric(
    owners_bornes[0], errors="coerce"
)
df["owners_max"] = pd.to_numeric(
    owners_bornes[1], errors="coerce"
)

# Estimation centrale, pratique pour les graphiques
df["owners_estimes"] = (
    df["owners_min"] + df["owners_max"]
) / 2

display(
    df[
        [
            "name",
            "price",
            "prix_euros",
            "owners",
            "owners_min",
            "owners_max",
            "owners_estimes",
        ]
    ].head(10)
)
```

```python

# Préparer le fichier final pour l'analyse
df_final = df.copy()

# Supprimer la colonne entièrement vide
df_final = df_final.drop(columns=["score_rank"])

# Enregistrer le fichier de travail mis à jour
df_final.to_csv(
    "gamescope_catalogue_jeux.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Fichier de travail mis à jour.")
print("Nombre de lignes :", len(df_final))
print("Nombre de colonnes :", len(df_final.columns))
print("\nColonnes finales :")
print(df_final.columns.tolist())
```
